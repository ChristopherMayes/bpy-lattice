````python
from pytao import Tao
import os

from bpy_lattice import Envelope

from bpy_lattice.interfaces.bmad import beam_envelope_from_dump

from bpy_lattice.blend import add_envelopes_to_blender
````

# Parameters

````python
INIT_FILE = "$ACC_ROOT_DIR/bmad-doc/tao_examples/optics_matching/tao.init"

DUMP_FILE = "beam_dump.h5"
````

# Make beam dump 

````python
tao = Tao(init_file=INIT_FILE, plot="mpl")
````

````python
tao.cmd(f"set beam dump_file = {DUMP_FILE}")
tao.cmd("set beam dump_at = *")
tao.cmd("set global track_type = beam;set global track_type = single")
````

# Make Envelope

````python
%%time
envelope = beam_envelope_from_dump(tao, DUMP_FILE)
````

````python
# 100x scale
envelope100 = beam_envelope_from_dump(tao, DUMP_FILE, scale=100)
````

# Serialize to JSON

````python
envelope100.to_json("envelope.json")
````

# Load from JSON and add to Blender

````python
envelope2 = Envelope.from_json("envelope.json")
````

````python
add_envelopes_to_blender([envelope2])
````
